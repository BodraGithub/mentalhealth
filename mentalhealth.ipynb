```python

# MENTAL HEALTH IN TECH SURVEY - EDA

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 12

print("Libraries imported successfully!")
```

```python

# 1. LOAD DATASET

df = pd.read_csv('survey.csv')
print(f"Dataset Shape: {df.shape}")
print(f"\nFirst 5 rows:")
df.head()
```

```python
# 2. DATA UNDERSTANDING


print("=" * 50)
print("DATASET INFORMATION")
print("=" * 50)
print(df.info())

print("\n" + "=" * 50)
print("STATISTICAL SUMMARY")
print("=" * 50)
df.describe(include='all')
```

```python
# Check missing values
print("=" * 50)
print("MISSING VALUES")
print("=" * 50)
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Percentage': missing_pct
}).sort_values('Missing Count', ascending=False)
print(missing_df[missing_df['Missing Count'] > 0])
```

```python
# Check duplicates
print(f"Duplicate rows: {df.duplicated().sum()}")
```

```python
# 3. DATA CLEANING & WRANGLING

df_clean = df.copy()

# 3.1 Clean Age column
print("Age statistics before cleaning:")
print(df_clean['Age'].describe())

# Remove unrealistic ages
df_clean = df_clean[(df_clean['Age'] >= 15) & (df_clean['Age'] <= 75)]
print(f"\nAfter age filtering: {df_clean.shape}")
```

```python
# 3.2 Clean Gender column
print("Original unique genders:")
print(df_clean['Gender'].unique())
```

```python
# Standardize gender
def clean_gender(gender):
    if pd.isna(gender):
        return 'Unknown'
    gender = str(gender).lower().strip()

    male_terms = ['male', 'm', 'man', 'cis male', 'mal', 'male-ish', 'maile',
                  'malr', 'cis man', 'make', 'guy (-ish) ^_^', 'male leaning androgynous']
    female_terms = ['female', 'f', 'woman', 'cis female', 'femake', 'female ',
                    'cis-female/femme', 'female (cis)', 'femail']
    trans_terms = ['trans-female', 'trans woman', 'female (trans)', 'trans woman']

    if gender in male_terms or 'male' in gender and 'female' not in gender and 'trans' not in gender:
        return 'Male'
    elif gender in female_terms or ('female' in gender and 'trans' not in gender and 'male' not in gender):
        return 'Female'
    elif gender in trans_terms or 'trans' in gender:
        return 'Trans'
    else:
        return 'Other'

df_clean['Gender'] = df_clean['Gender'].apply(clean_gender)
print("\nCleaned gender distribution:")
print(df_clean['Gender'].value_counts())
```

```python
# 3.3 Handle missing values
# Fill self_employed with mode
df_clean['self_employed'].fillna('No', inplace=True)

# Fill work_interfere with 'Don't know'
df_clean['work_interfere'].fillna("Don't know", inplace=True)

# Drop comments column for analysis
df_clean.drop('comments', axis=1, inplace=True)

print("Missing values after cleaning:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
# 3.4 Create Age groups
bins = [15, 25, 35, 45, 55, 75]
labels = ['15-24', '25-34', '35-44', '45-54', '55+']
df_clean['Age_Group'] = pd.cut(df_clean['Age'], bins=bins, labels=labels)

print("Age group distribution:")
print(df_clean['Age_Group'].value_counts().sort_index())
```

```python
# Final cleaned dataset info
print(f"\nFinal cleaned dataset shape: {df_clean.shape}")
df_clean.head()
```

```python
# 6. MULTIVARIATE ANALYSIS


# Chart 13: Treatment rate by Country and Gender (Top 5 countries)
top_countries = df_clean['Country'].value_counts().head(5).index
df_top = df_clean[df_clean['Country'].isin(top_countries)]

treatment_by_country_gender = df_top.groupby(['Country', 'Gender'])['treatment'].apply(
    lambda x: (x == 'Yes').mean() * 100
).reset_index(name='Treatment_Rate')

fig = px.bar(treatment_by_country_gender, x='Country', y='Treatment_Rate',
             color='Gender', barmode='group',
             title='Treatment Rate by Country and Gender (Top 5 Countries)',
             color_discrete_sequence=px.colors.qualitative.Set2)
fig.update_layout(yaxis_title='Treatment Rate (%)')
fig.show()
```

```python
# Chart 18: Correlation Heatmap
from sklearn.preprocessing import LabelEncoder

df_encoded = df_clean.copy()
le = LabelEncoder()
categorical_cols = df_encoded.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))

# Select numerical columns for correlation
corr_cols = ['Age', 'family_history', 'treatment', 'work_interfere',
             'benefits', 'care_options', 'wellness_program', 'seek_help',
             'anonymity', 'mental_health_consequence', 'phys_health_consequence',
             'coworkers', 'supervisor', 'mental_health_interview',
             'mental_vs_physical', 'obs_consequence']

corr_matrix = df_encoded[corr_cols].corr()

fig = px.imshow(corr_matrix, text_auto='.2f', aspect='auto',
                title='Correlation Heatmap of Key Variables',
                color_continuous_scale='RdBu_r', zmin=-1, zmax=1)
fig.update_layout(width=900, height=800)
fig.show()
```

```python
# Chart 19: Pair Plot
pair_cols = ['Age', 'family_history', 'treatment', 'benefits', 'care_options']
pair_df = df_encoded[pair_cols].copy()
pair_df['treatment'] = df_clean['treatment']

fig = px.scatter_matrix(pair_df, dimensions=pair_cols[:4], color='treatment',
                        title='Pair Plot of Key Variables',
                        color_discrete_sequence=px.colors.qualitative.Set2)
fig.update_layout(width=900, height=900)
fig.show()
```

```python
# Chart 20: Stacked Bar - Mental vs Physical Health by Company Size
mental_vs_physical = pd.crosstab(df_clean['no_employees'], df_clean['mental_vs_physical'],
                                  normalize='index') * 100
fig = px.bar(mental_vs_physical, barmode='stack',
             title='Mental vs Physical Health Perception by Company Size (%)',
             color_discrete_sequence=px.colors.qualitative.Set3)
fig.update_layout(xaxis_title='Company Size', yaxis_title='Percentage (%)')
fig.show()
```
